# ***Stemming in NLP***
***Stemming is the process of reducing words to their root/base form
(stem) by removing prefixes or suffixes***

# ***Important: spaCy does not provide a built-in stemming algorithm.***
spaCy primarily uses lemmatization, If you specifically want
stemming while using spaCy, you can integrate
a stemmar such as NLTK's Porterstemmer or SnowballStemmer.

# 1. Stemming using NLTK

```python
!pip install nltk
```

```python
# Import PorterStemmer

from nltk.stem import PorterStemmer
```

```python
# 1.: Basic stemming

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "player"]

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 2. Stemming a sentence

text = "The students are studying different subjects."

words = text.split()

for word in words:
  print(word, "->", stemmer.stem(word))

words = ["playing", "played", "plays", "player"]

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 3. Words ending with "ing"

words = ["playing", "reading", "writing", "walking", "running"]

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 4. Words ending with "ed"

words = ["played", "walked", "talked", "connected", "completed"]

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 5. Words ending with "s"

words = ["cats", "dogs", "books", "cars", "students"]

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 6. Different forms of the same word

words = ["connect", "connected", "connecting", "connection", "connections"]

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 7. Stemming with a loop

words =["happy", "happier", "happiest", "happiness"]

for word in words:
  stem = stemmer.stem(word)
  print(f"{word:12} -> {stem}")
```

```python
# 8. Stemming words related to organize

words = ["organize", "organized", "organizing", "organization", "organizations"]

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 9. Stemming text using regular expressions

import re

text = "Students are learning Python and developing machine learning projects."

words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# 10. Created a stemmed sentence

text = "The players are playing football and practicing regulary."

words = text.lower().split()

stemmed_words = []

for word in words:
  stemmed_words.append(stemmer.stem(word))

stemmed_text = "".join(stemmed_words)
print("Original:")
print(text)

print("\nStemmed:")
print(stemmed_text)
```

# 2.Stemming with spaCy

Important Note
spaCy does not have a native stem() function like NLTK.
Therefore, if we want to demonstrate stemming in a spaCy workflow, we can use:
spaCy ---> Tokenization ---> NLTK Stemmer

```python
!pip install spacy nltk
```

```python
!python -m spacy download en_core_web_sm
```

```python
# Importing library
import spacy
from nltk.stem import PorterStemmer

nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
```

```python
# 1. Stemming spaCy tokens
text = "The students are studying."

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
# 2. Playing, played, plays

text = "I play football. He plays football. They played yesterday."

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
# 3. Reading-related words

text = "Reading readers read readable books."

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
# 4. Learning-related words

text = "Students learn learning learned concepts."

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
# 5. Connect-related words

text = "The system connects connected devices."

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
# 6. Stemming a complete paragraph

text = """
Data scientists analyze data.
They are analyzing datasets and developed useful models.
"""

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

```python
# 7. Remove punctuation before stemming

text = "Python is powerful, useful, and popular!"

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
# 8. Stemming with lowercase conversion

text = "Python Pythonic Programming Programs"

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    word = token.text.lower()
    print(word, "->", stemmer.stem(word))
```

```python
# 9. Create a stemmed sentence using spaCy

import spacy
from nltk.stem import PorterStemmer

nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()

text = "The children are playing games and enjoying themselves."

doc = nlp(text)
stemmed_words = []

for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))

stemmed_text = " ".join(stemmed_words)

print("Original:")
print(text)
print("\nStemmed:")
print(stemmed_text)
```
