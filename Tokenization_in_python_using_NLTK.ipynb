```python
# 1.: Word Tokenization

from nltk.tokenize import word_tokenize

text = "Python is a popular programming language."

tokens = word_tokenize(text)

print(tokens)
```

```python
# 2. Sentence tokenization

from nltk.tokenize import sent_tokenize

text = "Python is easy to learn. NLTK is usedful for NLP. NLP is used to process text."

sentences = sent_tokenize(text)

print(sentences)
```

```python
# 3. Word tokenization of multiple sentences

from nltk.tokenize import word_tokenize

text = """ Python is easy to learn.
NLTK provides may NLP tools.
Tokenization is an important NLP task."""

tokens = word_tokenize(text)

print(tokens)
```

```python
# 4. Tokenizing User Input

from nltk.tokenize import word_tokenize

text = input("Enter a sentence: ")

tokens = word_tokenize(text)

print("Tokens:",tokens)
```

```python
# 5. Tokenization with punctuation

text = "Hello, how are you? I am fine!"

tokens = word_tokenize(text)

print(tokens)
```

```python
# 6. Tokenization of numbers

text = "The price is 500 rupees."

tokens = word_tokenize(text)

print(tokens)
```

```python
# 7. Tokenization with Contractions

text = "I don't like this movie."

tokens = word_tokenize(text)

print(tokens)
```

```python
# 8. Sentence Tokenization + Word Tokenization

from nltk.tokenize import sent_tokenize, word_tokenize

text = """Python is easy to learn.
NLTK is a python library.
It is widely used for NLP."""

sentences = sent_tokenize(text)

for sentence in sentences:
  words = word_tokenize(sentence)
  print(words)
```

```python
# 9. Count Numbers of Tokens

from nltk.tokenize import word_tokenize

text = "Python is easy and python is powerful."

tokens = word_tokenize(text)

print("Tokens:", tokens)
print("Numbers of tokens:", len(tokens))
```

```python
# 10. Count only words

from nltk.tokenize import word_tokenize
import string

text = "Python is easy and powerful!"

tokens = word_tokenize(text)

words = [token for token in tokens if token not in string.punctuation]

print("All tokens:", tokens)
print("Words:", words)
print("Number of words:", len(words))
```

```python
# 11. Convert Tokens to lowercase

text = "Python Python PYTHON is easy."

tokens = word_tokenize(text)

lower_tokens = [word.lower() for word in tokens]

print("Original tokens:")
print(tokens)

print("\nLowercase tokens:")
print(lower_tokens)
```

```python
# 12. tokenization using wordpunct_tokenize()

from nltk.tokenize import wordpunct_tokenize

text = "Hello! Python is great. Isn't it?"

tokens = wordpunct_tokenize(text)

print(tokens)
```

```python
# 13. Tokenizing a paragraph

from nltk.tokenize import sent_tokenize, word_tokenize

text = """ Artificial Intelligence is changing the world.
Machine learning is a part of AI.
Natural Language Processing helps computers understand human language."""

sentences = sent_tokenize(text)

print("Number of sentences:", len(sentences))

for i, sentence in enumerate(sentences, start=1):
  print(f"\nSentence {i}:")
  print(sentence)

  tokens = word_tokenize(sentence)

  print("Tokens:", tokens)
```

```python
# 15. Complete Tokenization Program

import nltk
import string
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

text = """Python is a powerful programming language.
It is widely used in Data Science, Machine Learning, and NLP!"""

# Step 1: Sentence Tokenization
sentences = sent_tokenize(text)

print("Sentences:")
for sentence in sentences:
  print(sentence)

# Step 2: Word tokenization
tokens = word_tokenize(text)

print("\nAll Tokens:")
print(tokens)

# Step 3: Remove Punctuation
words = [word for word in tokens if word not in string.punctuation]

print("\nTokens without punctuation:")

# Step 4: Convert to lowercase
lower_words = [word.lower() for word in words]

print("\nLowercase Tokens:")

# Step 5: Count Tokens
print("\Total tokens:", len(tokens))
print("Total words:", len(words))
```
