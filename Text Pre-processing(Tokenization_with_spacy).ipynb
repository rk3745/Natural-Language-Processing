# Tokenization Using spaCy in python
Tokenization is the process of breaking text into smaller units called tokens A Token can be a word, punctuation mark, number, or special symbol.

```python
# 1. Then import spaCy:

import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
# 2. Basic Tokenization

text = "Python is easy to learn."

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 3. Tokenization with token index
# Each token has an index indicating its position in the document.
# token.i gives the position/index of the token.

text = "Python is easy to learn."

doc = nlp(text)

for token in doc:
  print(token.i, token.text)
```

```python
# 4. Tokenization with token attributes
# spacy provides several useful attributes for each token.

text = "Python is easy to learn."

doc = nlp(text)
for token in doc:
  print(token.text, token.is_alpha, token.is_punct, token.is_space)
```

```python
# 5. Tokenization of a sentence with punctuation
text = "Hello how are you?"
doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 6. Tokenization of Numbers
text = "I bought 5 books for $100."

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 7. Tokenization of Email Addresses

text = "Contact me at neema@example.com"

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 8. Tokenization of URLs

text = "Visit https://www.example.com for more information."

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 9. Tokenization of Contractions

text = "i don't like python."

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 10. Tokenization of Hyphenated words

text = "Data-driven decision-making is important."

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 11. Removing punctuation after tokenization
text ="python is powerful, easy, and popular!"
doc = nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

```python
# 12. Getting only alphabetic token
# we can use is_alpha to select only alphabetic token.
text ="python 3 is easy to learn!"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text)
```

```python
# 13. Converting Tokens into a list

text = "python is easy to learn popular."

doc = nlp(text)

tokens = [token.text for token in doc]

print(tokens)
```

```python
# 14. Tokenization of Multiple sentences

text = "python is easy to learn. It is widely used in Data Science."

doc = nlp(text)

for token in doc:
  print(token.text)
```

```python
# 15. Tokenization and sentence segmentation
# spaCy can also identify individual sentences.

text = "python is easy to learn. It is popular in Data Science."

doc = nlp(text)

for sentence in doc.sents:
  print("Sentence:", sentence.text)

  for token in sentence:
    print(token.text)

  print()
```

```python
# 16. Tokenization from a text file

with open ("/content/# make a class name Bank_account ha.txt", "r", encoding="utf-8") as file:
  text = file.read()

doc = nlp(text)

for token in doc:
    print(token.text)
```

```python
# 17. Complete Example

text = "Python is easy to learn! I have 5 years of Experience."

doc = nlp(text)

print("Token\tAlpha\tNumber\tPunctuation")

for token in doc:
  print(token.text,"\t", token.is_alpha, "\t", token.like_num, "\t", token.is_punct)
```
