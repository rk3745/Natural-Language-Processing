# Lemmatization in NLP using Python

lemmatization is the process of converting a word
into its base or dictionary form, called a lemma.

```python
# 1: lemmatization using spaCy

!pip install spacy
!python -m spacy download en_core_web_sm
```

```python
# Import spacy and load the English model:

import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
# Basic Lemmatization

text = "The boys are playing in the gardens."

doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

```python
# 2. Lemmatizing Plural Nouns

text = "Cats dogs houses cars"

doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

```python
# 3. Lemmatizing Verbs

text = "playing played plays"

doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

```python
# 4. Lemmatizing Different verb forms

text = "I am eating. She ate dinner. They have eaten food."

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", token.lemma_)
```

```python
# 5. Lemmatization with a sentence

text = "The students were studying their lessons."

doc = nlp(text)

lemmatized_text = " ".join(token.lemma_ for token in doc)

print(lemmatized_text)
```

```python
# 6. Lemmatization of Irregular words

text = "better best went gone mice children"

doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

```python
# 7. Lemmatization Based on part of speech

text = "The children are running quickly."

doc = nlp(text)

for token in doc:
  print(token.text, "POS:", token.pos_, "Lemma:", token.lemma_)
```

```python
# 8. Remove Punctuation and Lemmatize

text = "Dogs are barking, cats are sleeping!"

doc = nlp(text)

words = []

for token in doc:
  if not token.is_punct:
    words.append(token.lemma_)
print(words)
```

```python
# 9. Lemmatization with stopword Removal

text = "The students are learning data science"

doc = nlp(text)

words = []

for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)

print(words)
```

```python
# 10: Create a lemmatization function

def lemmatized_text(text):
  doc = nlp(text)

  result = []

  for token in doc:
    if not token.is_stop and not token.is_punct:
      result.append(token.lemma_)

  return " ".join(result)

text = "The students were studying different subjects."

result = lemmatized_text(text)

print(result)
```

# ***Part 2 : Lemmatization using NLTK***
NLTK provides lemmatization primarily through WordNetLemmatize

```python
!pip install nltk
```

```python
import nltk

nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
# 1. Basic NLTK Lemmatization

words = ["cats", "dogs", "cars", "houses"]

for word in words:
  print("word", "->", lemmatizer.lemmatize(word))
```

```python
# 2. Lemmatizing verbs

words = ["playing","played","plays"]

for word in words:
  print("word", "->", lemmatizer.lemmatize(word, pos="v"))
```

```python
# 3. Lemmatizing nouns

words = ["children", "mice", "geese", "men", "women"]

for word in words:
  print("word", "->", lemmatizer.lemmatize(word, pos="n"))
```

```python
# 4. lemmatizing Adjectives

words = ["better", "best", "faster", "happier"]

for word in words:
  print("word", "->", lemmatizer.lemmatize(word, pos="a"))
```

```python
# 5. Lemmatizing adverbs

words = ["quickly", "slowly", "happily"]

for word in words:
  print("word", "->", lemmatizer.lemmatize(word, pos="r"))
```

```python
# 6. Irregular Verb Lemmatization

words = ["went", "gone", "going", "eaten", "ate"]

for word in words:
  print("word", "->", lemmatizer.lemmatize(word, pos="v"))
```

```python
# 7. Compare Noun and Verb Lemmatization

word = "meeting"

print("As noun :", lemmatizer.lemmatize(word, pos="n"))
print("As verb :", lemmatizer.lemmatize(word, pos="v"))
```

```python
# 8. Lemmatizing a List of words

words = ["studies", "studying", "studied", "students", "books"]

for word in words:

  if word.endswith("ing") or word.endswith("ed"):
    lemma = lemmatizer.lemmatize(word, pos = "v")
  else:
    lemma = lemmatizer.lemmatize(word, pos = "n")

  print(word, "->", lemma)
```

```python
# 9. Lemmatization with tokenization

from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
text = "The students are studying books."

tokens = word_tokenize(text)

for token in tokens:
  lemma = lemmatizer.lemmatize(token)
  print(token, "->", lemma)
```

```python
# 10: Complete NLTK Lemmatization with POS
import nltk
from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("wordnet")
nltk.download("omw-1.4")
lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(word):
  tag = nltk.pos_tag([word])[0][1]

  if tag.startswith("J"):
    return wordnet.ADJ
  elif tag.startswith("V"):
    return wordnet.VERB
  elif tag.startswith("N"):
    return wordnet.NOUN
  elif tag.startswith("R"):
    return wordnet.ADV
  else:
    return wordnet.NOUN

text = "The children are playing in the garden"
tokens = word_tokenize(text)
for token in tokens:
  if word.isalpha():
    pos = get_wordnet_pos(token)
    lemma = lemmatizer.lemmatize(token,pos=pos)
    print(token,"->",lemma)
```
