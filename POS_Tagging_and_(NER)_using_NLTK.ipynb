# ***PART-A : (POS Tagging using NLTK)***

```python
# 1. Basic POS Tagging

sentence = "Python is easy to learn."

words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

print(pos_tags)
```

```python
# 2. POS Tagging of a simple sentence

sentence = "The Student reads a book."

words = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(words)

for word, tag in tags:
  print(word, "->", tag)
```

```python
# 3. Identifying Nouns

sentence = "The teacher explained machine learning."

words = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(words)

for word, tag in tags:
   if tag.startswith("NN"):
    print(word)
```

```python
# 4. Identifying Verbs

sentence = "Students study Python and develop projects."

words = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(words)

for word, tag in tags:
   if tag.startswith("VB"):
    print(word, "->", tag)
```

```python
# 5. Identifying Adjectives

sentence = "Python is a powerful and popular language."

words = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(words)

for word, tag in tags:
   if tag.startswith("JJ"):
    print(word, "->", tag)
```

```python
# 6. Identifying Adverbs

sentence = "The student quickly completed the assignment."

words = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(words)

for word, tag in tags:
   if tag.startswith("RB"):
    print(word, "->", tag)
```

```python
# 7. POS Tagging with multiple Sentences

text = """
Python is powerful.
Students learn machine learning.
Data Scientists analyze data.
"""

sentences = nltk.sent_tokenize(text)

for sentence in sentences:
  words = nltk.word_tokenize(sentence)
  tags = nltk.pos_tag(words)

  print("\nSentence:", sentence)
  print(tags)
```

```python
# 8. Count POS Tags

from collections import Counter
import nltk

sentence = "The data scientist analyzes large datasets using Python."

words = nltk.word_tokenize(sentence)
tags = nltk.pos_tag(words)

pos_count = Counter(tag for word, tag in tags)

print(pos_count)
```

```python
# 9. Basic NER

sentence = "Barack Obama was born in Hawaii."

words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

tree = nltk.ne_chunk(pos_tags)

print(tree)
```

```python
# 10. Extract Person Names

sentence = "Sundar Pichai is the CEO of Google."

words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

tree = nltk.ne_chunk(pos_tags)

for chunk in tree:
  if hasattr(chunk, "label"):
    if chunk.label() == "PERSON":
      print("Person:", " ".join(c[0] for c in chunk))
```

```python
# 11. Extract Organizations

sentence = "Microsoft and Google are technology companies."

words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

tree = nltk.ne_chunk(pos_tags)

for chunk in tree:
  if hasattr(chunk, "label"):
    if chunk.label() == "ORGANIZATION":
      print("ORGANIZATION:", " ".join(c[0] for c in chunk))
```

```python
# 12. Extract Geographical Entities

sentence = "Rahul travelled from Delhi to Mumbai."

words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

tree = nltk.ne_chunk(pos_tags)

for chunk in tree:
  if hasattr(chunk, "label"):
    if chunk.label() in ["GPE", "LOCATION"]:
      print("Place :", " ".join(c[0] for c in chunk))
```

```python
# 13. Extract all named Entities

sentence = """
Elon Musk founded SpaceX.
SpaceX is headquarted in California.
"""
sentences = nltk.sent_tokenize(sentence)

for sent in sentences:

  words = nltk.word_tokenize(sentence)
  pos_tags = nltk.pos_tag(words)

  tree = nltk.ne_chunk(pos_tags)

  print("\nsentence:", sent)
  print(tree)
```

```python
# 14. Create a function for NER

import nltk

def extract_entities(sentence):

  words = nltk.word_tokenize(sentence)
  pos_tags = nltk.pos_tag(words)

  tree = nltk.ne_chunk(pos_tags)

  entities = []

  for chunk in tree:
    if hasattr(chunk, "label"):
      entity = " ".join(word for word, tag in chunk)
      entities.append((entity, chunk.label()))

  return entities

sentence = "Tim Cook works at Apple in California."


print(extract_entities(sentence))
```

```python
# POS + NER Together

text = """
Sundar Pichai is the CEO of Google.
He lives in California.
Google was founded in 1998.
"""

sentences = nltk.sent_tokenize(text)

for sentence in sentences:
  print("\nsentence:", sentence)

# Tokenization
words = nltk.word_tokenize(sentence)

# POS Tagging
pos_tags = nltk.pos_tag(words)

print("nPOS Tags:")
for word, tag in pos_tags:
  print(f"{word:15} {tag}")

# NER
tree = nltk.ne_chunk(pos_tags)

print(":\nNamed Entities:")

for chunk in tree:
  if hasattr(chunk, "label"):
    entity = " ".join(word for word, tag in chunk)
    print(entity, "->", chunk.label())
```
