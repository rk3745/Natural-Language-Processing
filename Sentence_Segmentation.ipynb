```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
#1 Basic Sentence segmentation
from nltk.tokenize import sent_tokenize
text = "Python is easy to learn . NLP is interesting! Do you like NLP"
sentences = sent_tokenize(text)
for sentence in sentences:
  print(sentence)
```

```python
#2 Counting Sentences
text = """
Python is a programming language.
NLP stands for Natural Language Processing.
It is used to process human language.
"""
sentences = sent_tokenize(text)
print(len(sentences))
```

```python
#3 Display sentence number
text = "I am learning Python. Python is useful for NLP. NLP is a branch of AI"
sentences = sent_tokenize(text)
for i , sentence in enumerate(sentences,start=1):
  print(f"Sentence{i}: {sentence}")
```

```python
#4 Sentence Segmentation with Different Punctuation
text = "Are you learning Python? Yes! Python is useful for NLP."
sentences = sent_tokenize(text)
for sentence in sentences:
  print(sentence)
```

```python
#5 Sentence Sefmentation From Multiple Lines
text = """Python is easy to learn.
It is widely used in Data Science.
NLP is interesting.
Do you like NLP?"""
sentences = sent_tokenize(text)
for sentence in sentences:
  print(sentence)
```

```python
#6 Sentence segmentation with Abbreviations
text = "Dr. Sharma teaches NLP. He uses Python in his classes"
sentences = sent_tokenize(text)
for sentence in sentences:
  print(sentence)
```

```python

!pip install spacy
!python -m spacy download en_core_web_sm
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
#7 Sentence segmentation using spacy
text = "Python is easy to learn. NLP is interesting ! Do you like NLP?"
doc = nlp(text)
for sentence in doc.sents:
  print(sentence.text)
```

```python
#8 Display sentence Number using spacy
text = "Python is powerful. It is used in Data science. NLP is one of its applications"
doc = nlp(text)
for i , sentence in enumerate(doc.sents,start=1):
  print(f"Sentence{i}: {sentence.text}")
```

```python
#9 Get numbers of sentences using spacy
text= "Python is easy. NLP is interesting . Machine learning is useful."
doc= nlp(text)
sentences = list(doc.sents)
print("number of sentences:",len(sentences))
```

```python
#10 Access Individual Sentences
text = "Python is easy. NLP is interesting. AI is powerful."
doc = nlp(text)
sentences = list(doc.sents)
print("First Sentence:",sentences[0])
print("Second Sentence:",sentences[1])
print("Third Sentence:",sentences[2])
```

```python
#11 Sentence segmemntation using spacy sentencizer
# spacy can perform simple rule-based sentence
#segmentation using sentencizer without loading a large language model.
nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")
text = "Python is easy. NLP is interesting. AI is powerful."
doc = nlp(text)
sentences = list(doc.sents)
for sentence in sentences:
  print(sentence.text)
```

```python
#12 Custom sentence Segmentation with spacy
# you can alo define your sentence boundaries
text = "Python is easy. NLP is interesting. Data Science is useful."
doc = nlp(text)
for sentence in doc.sents:
  print(sentence.text)
```

```python
#13 Sentence segmentation + Tokenization using spacy
text = "Python is easy to learn . NLP is interesting."
doc = nlp(text)
for sentence in doc.sents:
   print("\nSentence:",sentence.text)
   print("\nTokens:")
   for token in sentence:
     print(token.text)
```
